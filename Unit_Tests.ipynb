```python
from datetime import datetime

from habit_tracker import (
    Habit,
    get_all_habits,
    get_habits_by_periodicity,
    get_habit_longest_streak,
    get_longest_streak,
    save_habits,
    create_habit_from_data,
    load_habits,
    delete_habit
)
```

```python
import unittest
import os
import tempfile

class TestHabit(unittest.TestCase):

    def test_create_habit(self):
        habit = Habit("Exercise", "daily")

        self.assertEqual(habit.habit_name, "Exercise")
        self.assertEqual(habit.periodicity, "daily")
        self.assertEqual(habit.completed_dates, [])

    def test_check_off(self):
        habit = Habit("Read", "daily")
    
        habit.check_off()
    
        self.assertEqual(len(habit.completed_dates), 1)

    def test_daily_streak(self):
        habit = Habit("Exercise", "daily")
    
        habit.completed_dates = [
            datetime(2026, 10, 1),
            datetime(2026, 10, 2),
            datetime(2026, 10, 3),
            datetime(2026, 10, 5)
        ]
    
        result = get_habit_longest_streak(habit)
    
        self.assertEqual(result, 3)

    def test_weekly_streak(self):
        habit = Habit("Clean room", "weekly")
    
        habit.completed_dates = [
            datetime(2026, 9, 21),
            datetime(2026, 9, 30),
            datetime(2026, 10, 9),
            datetime(2026, 10, 20)
        ]
    
        result = get_habit_longest_streak(habit)
    
        self.assertEqual(result, 3)

    def test_habits_by_periodicity(self):
        habit1 = Habit("Exercise", "daily")
        habit2 = Habit("Read", "daily")
        habit3 = Habit("Clean room", "weekly")
    
        test_habits = [habit1, habit2, habit3]
    
        result = get_habits_by_periodicity(test_habits, "weekly")
    
        self.assertEqual(result, [habit3])

    def test_longest_streak(self):
        habit1 = Habit("Exercise", "daily")
        habit2 = Habit("Read", "daily")
    
        habit1.completed_dates = [
            datetime(2026, 10, 1),
            datetime(2026, 10, 2),
            datetime(2026, 10, 3)
        ]
    
        habit2.completed_dates = [
            datetime(2026, 10, 5),
            datetime(2026, 10, 6),
            datetime(2026, 10, 7),
            datetime(2026, 10, 8)
        ]
    
        test_habits = [habit1, habit2]
    
        result = get_longest_streak(test_habits)
    
        self.assertEqual(result, 4)

    def test_empty_streak(self):
        habit = Habit("Exercise", "daily")
    
        result = get_habit_longest_streak(habit)
    
        self.assertEqual(result, 0)

    def test_delete_habit(self):
        habit1 = Habit("Exercise", "daily")
        habit2 = Habit("Read", "daily")
    
        test_habits = [habit1, habit2]
    
        result = delete_habit(test_habits, "Exercise")
    
        self.assertTrue(result)
        self.assertEqual(test_habits, [habit2])

    def test_save_and_load(self):
        habit = Habit("Exercise", "daily")
        habit.completed_dates = [datetime(2026, 10, 1)]
    
        original_folder = os.getcwd()
    
        with tempfile.TemporaryDirectory() as test_folder:
            try:
                os.chdir(test_folder)
    
                save_habits([habit])
                loaded_habits = load_habits()
    
                self.assertEqual(len(loaded_habits), 1)
                self.assertEqual(loaded_habits[0].habit_name, "Exercise")
                self.assertEqual(loaded_habits[0].periodicity, "daily")
                self.assertEqual(
                    loaded_habits[0].completed_dates,
                    habit.completed_dates
                )
    
            finally:
                os.chdir(original_folder)
```

```python
unittest.main(argv=[''], exit=False)
```
