```python
from habit_tracker import (
    Habit,
    get_all_habits,
    get_habits_by_periodicity,
    get_habit_longest_streak,
    get_longest_streak,
    save_habits,
    load_habits,
    delete_habit
)
```

# CLI

```python
habits = load_habits()

while True:
    print("Habit Tracker")
    print("1. Create habit")
    print("2. Delete habit")
    print("3. Check off habit")
    print("4. Analyse habits")
    print("5. Exit")

    choice = input("Please choose an option: ")

    if choice == "1":
        print("You chose Create habit.")
        habit_name = input("What is the name of the habit? ")

        while True:
            periodicity = input("Is this a daily or weekly habit? ").lower()

            if periodicity not in ["daily", "weekly"]:
                print("Invalid periodicity. Please choose daily or weekly.")
            else:
                new_habit = Habit(habit_name, periodicity)
                habits.append(new_habit)
                print("Habit added successfully!")
                break

        save_habits(habits)
    
    elif choice == "2":
        print("You chose Delete habit.")
        habit_name = input("Which habit would you like to delete? ")
        habit_deleted = delete_habit(habits, habit_name)

        if habit_deleted:
            save_habits(habits)
    
    elif choice == "3":
        print("You chose Checkoff habit.")
        habit_name = input("Which habit would you like to check off? ")
    
        habit_found = False
    
        for habit in habits:
            if habit.habit_name.lower() == habit_name.lower():
                habit.check_off()
                habit_found = True
                save_habits(habits)
                break
    
        if not habit_found:
            print("Habit not found. Please try again.")


    elif choice == "4":
        print("1. View all habits")
        print("2. View habits by periodicity")
        print("3. View longest streak")
        print("4. View longest streak for a specific habit")
    
        analysis_choice = input("Please choose an option: ") 
    
        if analysis_choice == "1":
            get_all_habits(habits)
            print([habit.habit_name for habit in habits])
            
        elif analysis_choice == "2":
            periodicity = input("What periodicity would you like to analyse?: ").lower()
            matching_habits = get_habits_by_periodicity(habits, periodicity)
            print([habit.habit_name for habit in matching_habits])

        elif analysis_choice == "3":
            longest_streak = get_longest_streak(habits)
            print("Longest streak:", longest_streak)

        elif analysis_choice == "4":
            habit_name = input("Which habit would you like to analyse? ")
            habit_found = False

            for habit in habits:
                if habit.habit_name.lower() == habit_name.lower():
                    longest_streak = get_habit_longest_streak(habit)
                    print("Longest streak:", longest_streak)
                    habit_found = True
                    break

            if not habit_found:
                print("Habit not found. Please try again.")
    
    elif choice == "5":
        print("Goodbye!")
        break
        
    else:
        print("Invalid choice.")
```
